# LA Crime Data Visualizations

This notebook demonstrates quick exploratory plots using the **Crime Data from 2020 to Present** dataset.

We first load the cleaned dataframe via **`data_wrangling.load_and_clean()`**, then produce:

1. A time‑series plot showing the monthly count of reported crimes.
2. A bar chart of the 10 most common crime descriptions.

Feel free to build on these starting points for deeper analysis!


```

from pathlib import Path
import matplotlib.pyplot as plt
import pandas as pd

# Import the helper from our script
from data_wrangling import load_and_clean

RAW_CSV = Path('/mnt/data/Crime_Data_from_2020_to_Present (1).csv')  # adjust if needed
df = load_and_clean(RAW_CSV)

print(f"Loaded {len(df):,} rows")
df.head()
```

```

# --- Plot 1: Monthly crime counts ---
monthly = (
    df['occurrence_datetime']
    .dt.to_period('M')
    .value_counts()
    .sort_index()
)

plt.figure(figsize=(10, 4))
monthly.index = monthly.index.to_timestamp()
plt.plot(monthly.index, monthly.values)
plt.title("Monthly Reported Crimes in LA (2020‑present)")
plt.xlabel("Month")
plt.ylabel("Number of crimes")
plt.tight_layout()
plt.show()
```

```

# --- Plot 2: Top 10 crime categories ---
top10 = df['crm_cd_desc'].value_counts().head(10)

plt.figure(figsize=(10, 5))
top10.plot.bar()
plt.title("Top 10 Reported Crime Types")
plt.ylabel("Frequency")
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```
